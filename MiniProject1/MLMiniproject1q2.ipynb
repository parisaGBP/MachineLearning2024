#Feature extraction

```python
import numpy as np

# For each feature we used the given formula
#1. Peak
def calculate_peak(data):
    return np.max(data)

#2.crest_factor
def calculate_crest_factor(data):
    peak = calculate_peak(data)
    rms = np.sqrt(np.mean(np.square(data)))
    return peak / rms

#3.clearance_factor
def calculate_clearance_factor(data):
    peak = calculate_peak(data)
    return peak / (np.max(np.abs(data)))
#4.shape_factor
def calculate_shape_factor(data):
    rms = np.sqrt(np.mean(np.square(data)))
    mean = np.mean(data)
    return rms / mean

#5.mean
def calculate_mean(data):
    return np.mean(data)

#6. absolute_mean
def calculate_absolute_mean(data):
    return np.mean(np.abs(data))

#7.peak_to_peak
def calculate_peak_to_peak(data):
    return np.max(data) - np.min(data)

#8.impact_factor
def calculate_impact_factor(data):
    return np.max(np.abs(data))

n = 100
# A : Normal
Normal_features = []
for i in range(n):
    features = []
    data = NormalMatrix[:, i]
    features.append(calculate_peak(data))
    features.append(calculate_crest_factor(data))
    features.append(calculate_clearance_factor(data))
    features.append(calculate_absolute_mean(data))
    features.append(calculate_mean(data))
    features.append(calculate_shape_factor(data))
    features.append(calculate_peak_to_peak(data))
    features.append(calculate_impact_factor(data))
    Normal_features.append(features)

# B : fault
Fault_features = []
for i in range(n):
    features = []
    data = FaultMatrix[:, i]
    features.append(calculate_peak(data))
    features.append(calculate_crest_factor(data))
    features.append(calculate_clearance_factor(data))
    features.append(calculate_absolute_mean(data))
    features.append(calculate_mean(data))
    features.append(calculate_shape_factor(data))
    features.append(calculate_peak_to_peak(data))
    features.append(calculate_impact_factor(data))
    Fault_features.append(features)

Normal_features = np.array(Normal_features)
fault_features = np.array(Fault_features)

# print("Normal Data features:")
# print(Normal_features)

# print("Fault Data features:")
# print(fault_features)
```

Logistic Regression

```python
#Logistic Regression (from Scratch)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def logistic_regression(x, w):
    y_hat = sigmoid(x @ w)
    return y_hat

#Binary Cross Entropy (BCE)

def bce(y, y_hat):
    loss = -(np.mean(y*np.log(y_hat) + (1-y)*np.log(1-y_hat)))
    return loss
```

Gradient

```python
#Gradient
def gradient(x, y, y_hat):
    grads = (x.T @ (y_hat - y)) / len(y)
    return grads


#Gradient Descent

def gradient_descent(w, eta, grads):
    w = w - eta*grads
    # w -= eta*grads
    return w
```

Accuracy

```python
#Accuracy

def accuracy(y, y_hat):
    acc = np.sum(y == np.round(y_hat)) / len(y)
    return acc

```

```python
import pandas as pd
import matplotlib.pyplot as plt
#hyper parameter
m = 7
w = np.random.randn(m+1, 1)
print(w.shape)
eta = 0.01
n_epochs = 2000

# X_train_minmax = np.hstack((np.ones((len(X_train_minmax), 1)), X_train_minmax))
print(X_train_minmax.shape)

error_hist = []

for epoch in range(n_epochs):
    # predictions
    y_hat = logistic_regression(X_train_minmax, w)

    # loss
    e = bce(y_train, y_hat)
    error_hist.append(e)

    # gradients
    grads = gradient(X_train_minmax, y_train, y_hat)

    # gradient descent

    w = gradient_descent(w, eta, grads)

    if (epoch+1) % 1 == 0:
        print(f'Epoch={epoch}, \t E={e:.4},\t w={w.T[0]}')

```
