# Data Generation

```python
import numpy as np
from sklearn.datasets import make_classification
X, y = make_classification(n_samples=1000, n_features=3, n_redundant=0, n_clusters_per_class=1, class_sep=2, n_classes=4, random_state=12)
X.shape, y.shape
```

```python
import matplotlib.pyplot as plt

print(X.shape, y.shape)
colors = np.array(['blue', 'red', 'yellow', 'green'])
plt.scatter(X[:, 0], X[:, 1], c=colors[y])
```

# Test and train data

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, SGDClassifier
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

# SGDClassifier

```python
from sklearn.linear_model import SGDClassifier
```

```python
model = SGDClassifier(loss='log_loss', random_state=27)
model.fit(x_train, y_train)
```

```python
model.score(x_train, y_train)
```

```python
model.score(x_test, y_test)
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.svm import SVC
pca = PCA(n_components = 2)
X1 = pca.fit_transform(X)
clf = SVC(C=100,gamma=0.0001)
clf.fit(X1, y)

colors = np.array(['blue', 'orange', 'green', 'red'])
plt.scatter(X1[:, 0], X1[:, 1], c=colors[y])
```

```python
from mlxtend.plotting import plot_decision_regions
from sklearn.decomposition import PCA
pca = PCA(n_components = 2)
X1 = pca.fit_transform(X)
clf=model
clf.fit(X1, y)
plot_decision_regions(X1, y, clf=model)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('/content/data1.csv')
df
```

```python
X = df[['x', 'y']].values
y = df[['z']].values
# X, y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, SGDClassifier
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
model = SGDClassifier(loss='log_loss', random_state=27)
model.fit(x_train, y_train)
```

```python
model.score(x_train, y_train)
```

```python
model.score(x_test, y_test)
```

```python
x1_min, x2_min = X.min(0)
x1_max, x2_max = X.max(0)

n = 500
x1r = np.linspace(x1_min, x1_max, n)
x2r = np.linspace(x2_min, x2_max, n)

x1m, x2m= np.meshgrid(x1r, x2r) # zoj noghte
print(x1m.shape, x2m.shape)
Xm = np.stack((x1m.flatten(), x2m.flatten()), axis=1)

ym = model.decision_function(Xm)

# ym.shape
import matplotlib.pyplot as plt

# Assign colors to classes
# colors = np.array(['blue', 'red'])
plt.scatter(X[:, 0], X[:, 1], c=y)


plt.contour(x1m, x2m, ym.reshape(x1m.shape), levels=[-1,0,1])
plt.show()
```
